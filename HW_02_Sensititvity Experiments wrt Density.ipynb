## A simple 0-dimensional climate model for global mean temperature

```python
# load modules
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# model parameter
dt      = 86400.0          # seconds in 24 hours
secday  = 86400.0          # seconds in 24 hours
secyear = 86400.0*360.0    # seconds in 1 year
maxtime = 20.0*secyear     # seconds in 20 years
itmax   = int(maxtime/dt)  # number of timesteps

pi      = math.pi

eps     = 0.61             # effective emissivity --> Closest approximation and residual to replicate real world scenario & simulate green house effect
sigma   = 5.67*(10**-8)    # Stefan-Boltzmann constant in W m-2 K-4
alpha   = 0              # planetary albedo (This could be the value for 70 m depth ocean)
solar   = 1370.0           # annual average value of solar constant (S) in W m-2
```

```python
# Physical Constants 
# Translate energy balance
# Study temperature evolution
# 12 degrees - Pre Industrial Global Mean Temperature
# Aqua Planet in this Block


H       = 70.0             # depth of the fluid column in m --> Ocean here is shallower   
rho_seawater = 1025             # density of (seawater) fluid in Kg m-3
rho_freshwater = 1000 # density of (freshwater) fluid in Kg m-3
rho_ice = 917 # density of ice in Kg m-3
rho_methane = 420 # density of methane in Kg m-3
rho_mercury = 13600 # density of mercury in Kg m-3
rho_ethanol = 789 # density of mercury in Kg m-3

cp      = 4000.0           # specific heat of fluid in J Kg-1 K-1 --> Actual value is lower


ce_seawater      = cp*rho_seawater*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 
ce_freshwater      = cp*rho_freshwater*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 
ce_ice     = cp*rho_ice*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 
ce_methane     = cp*rho_methane*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 
ce_mercury    = cp*rho_mercury*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 
ce_ethanol    = cp*rho_ethanol*H         # heat capacity in J K-1 m-2 --> To calculate actual heat capacity 



T_ini_seawater   = 285.15           # should be same in the fortran version for comparison
T_ini_freshwater   = 285.15           # should be same in the fortran version for comparison
T_ini_ice   = 285.15           # should be same in the fortran version for comparison
T_ini_methane   = 285.15           # should be same in the fortran version for comparison
T_ini_mercury  = 285.15           # should be same in the fortran version for comparison
T_ini_ethanol  = 285.15           # should be same in the fortran version for comparison
```

```python
# initialization for Sea Water
time    = np.zeros(itmax)
T_abs_seawater   = np.zeros(itmax)
T_abs_seawater[0]   = T_ini_seawater

# time loop
for it in range(0,itmax-1):    
            
        # compute processes 
        shortwave_seawater    = 0.25*solar*(1.0-alpha) # Incoming
        longwave_seawater     = eps*sigma*(T_abs_seawater[it]**4) # Outgoing       

        # compute absolute temperature
        T_abs_seawater[it+1]  = T_abs_seawater[it] + dt*(shortwave_seawater - longwave_seawater)/ce_seawater # Numerical SOlution to Equations

        # advance time
        time[it+1] = time[it]+dt

# Convert T_abs to Celsius        
T_seawater    = T_abs_seawater - 273.15

# Convert time axis to years        
time = time/(secday*360)

```

```python
# initialization for Freshwater
time    = np.zeros(itmax)
T_abs_freshwater   = np.zeros(itmax)
T_abs_freshwater[0]   = T_ini_freshwater

# time loop
for it in range(0,itmax-1):    
            
        # compute processes 
        shortwave_freshwater    = 0.25*solar*(1.0-alpha) # Incoming
        longwave_freshwater     = eps*sigma*(T_abs_freshwater[it]**4) # Outgoing       

        # compute absolute temperature
        T_abs_freshwater[it+1]  = T_abs_freshwater[it] + dt*(shortwave_freshwater- longwave_freshwater)/ce_freshwater # Numerical Solution to Equations

        # advance time
        time[it+1] = time[it]+dt

# Convert T_abs to Celsius        
T_freshwater   = T_abs_freshwater - 273.15

# Convert time axis to years        
time = time/(secday*360)
```

## Plot

```python
# plot Temperature versus time
fig = plt.figure(figsize=(10,5))
ax  = fig.add_subplot(111)
plt.title('Sensitvity Tests with Different Densities')
plt.plot(time,T_seawater,linewidth=4.0,label='T_Seawater',color='black')
#plt.plot(time,T_freshwater,linewidth=2.0,label='T_Freshwater',color='red')
plt.plot(time,T_ice,linewidth=2.0,label='T_Ice',color='blue')
plt.plot(time,T_methane,linewidth=2.0,label='T_Methane',color='green')
#plt.plot(time,T_mercury,linewidth=2.0,label='T_Mercury',color='yellow')
#plt.plot(time,T_ethanol,linewidth=2.0,label='T_Ethanol',color='magenta')
plt.xlim(min(time),max(time))
plt.ylim(min(T_seawater)-1,max(T_seawater)+1)
plt.ylim(min(min(T_seawater), min(T_freshwater),min(T_ice),min(T_methane), min(T_mercury),min(T_ethanol))-1,max(max(T_seawater), max(T_freshwater),max(T_ice),max(T_methane), max(T_mercury), max(T_ethanol))+1)
ax.set_xticks(time[::360])
ax.set_xticklabels(time[::360])
plt.xlabel("Year")
plt.ylabel("T (°C)")
plt.legend(framealpha=1, loc='lower right', frameon=True)
plt.show()
```
